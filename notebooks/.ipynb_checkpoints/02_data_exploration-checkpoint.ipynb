```python
import os
from PIL import Image
import matplotlib.pyplot as plt

# Set your dataset base path correctly here
base = './data/Plant_leaf_diseases_dataset_with_augmentation'

# Check if the base directory exists
if not os.path.exists(base):
    print(f"❌ Dataset path not found: {base}")
else:
    # List only directories (the disease classes)
    classes = [d for d in os.listdir(base) if os.path.isdir(os.path.join(base, d))]

    # Plotting sample images
    fig, axes = plt.subplots(2, 3, figsize=(12, 6))
    for ax, cls in zip(axes.flatten(), classes[:6]):
        img_folder = os.path.join(base, cls)
        img_files = os.listdir(img_folder)

        if not img_files:
            continue  # skip if folder is empty

        img_file = img_files[0]
        img_path = os.path.join(img_folder, img_file)

        try:
            img = Image.open(img_path)
            ax.imshow(img)
            ax.set_title(cls)
            ax.axis('off')
        except Exception as e:
            print(f"⚠ Could not load image: {img_path}\nError: {e}")
    
    plt.tight_layout()
    plt.show()
```
